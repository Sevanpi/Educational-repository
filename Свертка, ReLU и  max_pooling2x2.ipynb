```python
import numpy as np
from scipy.signal import correlate2d

```

```python
#Функция свёртки
def conv2d(x, k, stride=1, padding=0):

  x = np.pad(x, padding)
  f = k.shape[0]
  n = (x.shape[0] - f) // stride + 1
  m = (x.shape[1] - f) // stride + 1
  out = np.zeros((n,m), dtype=int)
  for i in range(n):
    for j in range(m):
      window = x[i*stride : i*stride + f , j*stride : j*stride + f]
      out[i, j] = (window * k).sum()

  return out

#Функция активации
def ReLU(a):
  return np.maximum(0, a)

#Функция max_pooling'а для матриц 2х2
def max_pool(a, f=2):
  rows = a.shape[0]
  cols = a.shape[1]
  if (rows % 2 != 0) or (cols % 2 != 0): return
  out = np.zeros((rows//2, cols//2))
  lst = list()
  for i in range(0, rows, 2):
    for j in range(0, cols, 2):
      lst.append(np.max(a[i:i+f , j:j+f]))

  #out = np.array(lst).reshape(2,2)
  out = np.array(lst).reshape(rows//2, cols//2)
  return out
```

```python
#Сравнение нашей функции conv2d и correlate2d из scipy.signal
x = np.tile([0,0,1,1,0,0], (6,1))
k = np.array([[1,0,-1]] * 3)
cnv = conv2d(x,k)
print(cnv)
print(correlate2d(x, k, mode='valid'))
```

```python
#Проверка функции max_pool и ReLU
print(max_pool(ReLU(cnv)))
```
